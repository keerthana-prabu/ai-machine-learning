```python
import pandas as pd 

data = {
    "name" : ["A", "B" , "A", "D", "E", "F", "G"],
    "class" : [1, 2, 1, 1, 2, 3, 1],
    "marks" : [98, 94, 98, 76, 92, 85, 40],
    "Grade" : ["A", "A", "A", "C", "A", "B", "F"]
}

df = pd.DataFrame(data)

print(df.head())
print(df.columns)
print(df.info())
display(df.head())
print(df.describe())
print(df.tail(2))
print(df.sample(n=1))
print(df.shape)
print(df["name"])
print(df[["name", "Grade"]])
print(df.iloc[2])
print(df.iloc[1, 2])
print(df.iloc[0:1, 0:2])
print(df.loc[2])
print(df.loc[:, ["name", "marks"]])
print(df["marks"]>90)
print(df[df["marks"]>90])
print(df[(df["marks"]>90) & (df["class"] == 2)])
df.isnull()
df.isnull().sum()
df = df.dropna()
df["marks"] = df["marks"].fillna(df["marks"].mean())
df["class"] = df["class"].fillna(df["class"].mode()[0])
print(df.duplicated().sum())
print(df["class"].value_counts())
df["blah"] = df["marks"] / 100
print(df["blah"])
df = df.drop("blah", axis=1)
print(df.sort_values("marks", ascending=False))
print(df.groupby("class")["marks"].mean())
print(df.corr(numeric_only=True))
```
